```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Data Structure

I have managed to download a dataset of the reaction Lu-176 to Lu-177 via neutron capture and gamma emission, exported as a csv file

The data is structured like follows:

Column; DATA(B) 0.1 is the measured cross section, in Barns

Column; DATA-ERR(B) 0.911 is the associated cross section uncertainty in barns

Column; EN (EV) 1.1 is the incident neutron energy

this is what we are interested in, we want to plot cross section as a function of neutron energy?

```python
## Loading in the data
data = pd.read_csv('exfordat2.csv')

## Extracting the columns
cross_sections = data['DATA (B) 0.1']
cs_sigma = data['DATA-ERR (B) 0.911']
n_energy = data['EN (EV) 1.1']

## We will want to make this logarithm-ed

```

```python
## Making some plots

plt.loglog()
plt.grid(linewidth=1, alpha=0.5)
plt.scatter(n_energy, cross_sections, marker='.', color='black')
plt.xlabel('Incident Neutron Energy (eV)')
plt.ylabel('Measured Cross Section (Barns)')
plt.show()
```

```python
## Checking the data:

print("Cross section range:",np.nanmin(cross_sections),"to", np.nanmax(cross_sections))

print("CS Error Range:", np.nanmin(cs_sigma), "to", np.nanmax(cs_sigma))

## Fractional uncertainties
ratio = cs_sigma / cross_sections

print('Fractional uncertainty range:', np.nanmin(ratio), np.nanmax(ratio))
```
