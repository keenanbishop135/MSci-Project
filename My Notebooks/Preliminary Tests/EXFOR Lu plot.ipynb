```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data_widder = pd.read_csv('20437011.csv')

energy_widder = data_widder['EN (EV) 1.1']
cross_section_widder = data_widder['DATA (B) 0.1']

plt.figure()
plt.scatter(energy_widder, cross_section_widder, color='black', marker='.', zorder=2)
plt.grid(True, which='Both', zorder=1)

plt.xscale('log')
plt.yscale('log')

plt.show()
```

```python
data_roig = pd.read_csv('exfordat2.csv')

roig_cs = data_roig['DATA (B) 0.1']
#x4_cs_sigma = x4_data['DATA-ERR (B) 0.911']
roig_energy = data_roig['EN (EV) 1.1']

plt.figure()
plt.scatter(roig_energy, roig_cs, color='black', marker='o', zorder=2, label='O. Roig Data (2016)')
plt.scatter(energy_widder, cross_section_widder, color='red', marker='.', zorder=3, label='F. Widder Data (1975) ')
plt.grid(True, which='Both', zorder=1)

plt.xscale('log')
plt.yscale('log')
plt.xlim(0.009, 3)
plt.xlabel('Incident Neutron Energy (eV)')
plt.ylabel('Reaction Cross-Section (Barns)')
plt.title(r'$^{176}$Lu (n, $\gamma$) $^{177}$Lu')
plt.legend()

plt.show()
```
